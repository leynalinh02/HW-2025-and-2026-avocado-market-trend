# Add two files together (must be same structure)

```python
# use concatenate
# Read in the 2025 dataset as data2025 

# Read the dataset and store it as data2025
data2025=pd.read_csv('2025.csv',encoding='utf-8')
```

```python
# Verify AI memo
```

```python
# Total U.S. data for weeks 1-32
us2025 = data2025[
    (data2025['Geography'] == 'Total U.S.') &
    (data2025['Period'] >= 1) &
    (data2025['Period'] <= 32)
]

us2026 = data2026[
    (data2026['Geography'] == 'Total U.S.') &
    (data2026['Period'] >= 1) &
    (data2026['Period'] <= 32)
]
```

```python
# Compare average price and sales volume
summary2025 = us2025.groupby('Type').agg({
    'ASP Current Year': 'mean',
    'Total Bulk and Bags': 'sum'
})

summary2026 = us2026.groupby('Type').agg({
    'ASP Current Year': 'mean',
    'Total Bulk and Bags': 'sum'
})

print("2025")
print(summary2025)

print("\n2026")
print(summary2026)
```

```python
# Calculate percentage changes from 2025 to 2026
percent_change = ((summary2026 - summary2025) / summary2025) * 100

print("Percent Change from 2025 to 2026")
print(percent_change)
```

```python
# Compare organic and conventional price difference
price_difference_2025 = 1.740879 - 1.215869
price_difference_2026 = 1.564469 - 0.984290

print("2025 Price Difference:", round(price_difference_2025, 2))
print("2026 Price Difference:", round(price_difference_2026, 2))
```
